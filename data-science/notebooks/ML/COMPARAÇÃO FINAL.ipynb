# Análise Comparativa dos Modelos de Classificação

Este notebook tem como objetivo realizar uma análise comparativa entre os principais modelos de classificação testados no projeto, considerando o intervalo temporal de 2000 a 2008.

Após as etapas anteriores de exploração dos dados, definição dos rótulos ambientais, testes com diferentes intervalos temporais e ajustes nos algoritmos, foi identificado que o período entre 2000 e 2008 apresentou uma maior concentração de amostras críticas, sendo utilizado como base para a comparação final dos modelos.

Nesta etapa, serão comparados os melhores resultados obtidos pelos seguintes algoritmos:

- Random Forest
- LightGBM
- SVM
- Regressão Logística

A comparação será feita com base em métricas de desempenho, como acurácia, precisão, recall e F1-score, além da análise da matriz de confusão e do comportamento geral de cada modelo diante das classes ambientais.

O objetivo final é identificar qual modelo apresentou o melhor equilíbrio entre desempenho, capacidade de generalização e interpretação dos resultados, para que ele possa ser utilizado posteriormente no dashboard analítico do projeto.

# Comparação Final entre os Modelos

Os modelos avaliados foram:

- Random Forest

Para tornar a comparação mais justa, foi selecionada a melhor configuração de cada algoritmo, considerando o desempenho geral do modelo e o comportamento diante da classe **Não adequada**, que representa uma condição ambiental mais crítica.

A análise será feita com base nas seguintes métricas:

- **Accuracy**: indica o percentual geral de acertos do modelo.
- **Macro F1-score**: avalia o desempenho médio entre as classes, sem favorecer a classe majoritária.
- **F1-score da classe Não adequada**: mede o equilíbrio entre precisão e recall para a classe crítica.
- **Recall da classe Não adequada**: indica a capacidade do modelo de identificar corretamente os casos críticos.

Essa comparação será utilizada para apoiar a escolha do modelo mais adequado para uso posterior no dashboard analítico.

## Importância da Classe Não Adequada

Embora o problema estudado possua três classes ambientais distintas (**Adequada**, **Boa** e **Não adequada**), uma atenção especial foi direcionada à classe **Não adequada** durante a avaliação dos modelos.

Essa escolha está relacionada ao objetivo prático do projeto. Enquanto erros de classificação entre as classes *Adequada* e *Boa* tendem a representar pequenas variações na condição ambiental do corpo hídrico, falhas na identificação da classe *Não adequada* podem ocultar situações potencialmente críticas de qualidade da água.

Em aplicações reais de monitoramento ambiental, deixar de identificar corretamente um corpo hídrico em condição inadequada pode gerar interpretações equivocadas sobre o estado do ambiente, comprometendo processos de fiscalização, tomada de decisão e planejamento de ações corretivas.

Por esse motivo, além das métricas globais como Accuracy e Macro F1-score, foi realizada uma análise específica do desempenho dos algoritmos na classe **Não adequada**, observando principalmente os valores de Precision, Recall e F1-score.

Dessa forma, a avaliação dos modelos considera não apenas a capacidade geral de classificação, mas também a eficiência na identificação das situações ambientais de maior interesse para o projeto.

## LightGBM

O LightGBM apresentou o melhor equilíbrio geral entre as métricas avaliadas. O modelo alcançou a maior acurácia entre os algoritmos analisados e manteve um desempenho consistente nas três classes.

Além disso, apresentou valores superiores de Macro F1-score, indicando melhor capacidade de classificação em cenários com distribuição desigual entre as classes.

O desempenho observado sugere uma boa capacidade de generalização dos dados, tornando o LightGBM o candidato mais adequado para utilização no dashboard analítico do projeto.

## Random Forest

O Random Forest apresentou desempenho próximo ao LightGBM em diversas métricas, demonstrando boa capacidade de classificação geral.

Entretanto, apresentou valores inferiores para a classe Não adequada, especialmente em Recall e F1-score, indicando menor eficiência na identificação dos casos críticos.

Apesar disso, o modelo demonstrou estabilidade e resultados consistentes ao longo dos experimentos realizados.

## SVM

O SVM destacou-se principalmente pela elevada capacidade de identificação da classe Não adequada, alcançando o maior Recall entre todos os modelos avaliados.

Esse comportamento indica que o algoritmo foi capaz de detectar a maior parte dos casos críticos presentes no conjunto de teste.

Por outro lado, esse ganho ocorreu às custas de uma redução no desempenho geral e de dificuldades na classificação da classe Boa, resultando em menor equilíbrio entre as classes.

## Regressão Logística

A Regressão Logística apresentou os menores resultados entre os algoritmos avaliados.

Embora tenha conseguido identificar satisfatoriamente a classe Adequada, apresentou dificuldades na classificação das classes Boa e Não adequada, refletindo em menores valores de Recall e F1-score.

Esses resultados sugerem que modelos lineares possuem limitações para capturar os padrões presentes nos dados ambientais analisados.

# Escolha do Modelo Final

Considerando os resultados obtidos ao longo da análise comparativa, o algoritmo LightGBM foi selecionado como modelo final do projeto.

A decisão foi baseada principalmente nos seguintes fatores:

- Maior desempenho geral entre os modelos avaliados;
- Melhor equilíbrio entre as classes;
- Melhor Macro F1-score;
- Boa capacidade de identificação da classe Não adequada;
- Menor evidência de sobreajuste quando comparado a outros algoritmos.

Embora o SVM tenha apresentado maior Recall para a classe Não adequada, seu desempenho geral foi inferior ao do LightGBM, comprometendo o equilíbrio da classificação.

Dessa forma, o LightGBM foi considerado o modelo mais adequado para integração ao dashboard analítico e para futuras etapas de interpretação dos resultados.

## Evolução da Estratégia de Modelagem

Os resultados apresentados nesta etapa são fruto de um processo iterativo de análise e refinamento dos modelos de classificação.

Nas primeiras versões do problema, utilizando uma estratégia diferente de rotulagem e considerando todo o intervalo temporal disponível, o SVM apresentou os melhores resultados entre os algoritmos avaliados. Entretanto, análises posteriores revelaram dificuldades significativas na diferenciação entre algumas classes, especialmente em regiões onde havia forte sobreposição dos dados.

Diante desse cenário, foram realizados novos experimentos envolvendo técnicas de balanceamento, revisão da estratégia de rotulagem e análise da distribuição temporal das amostras. Esse processo permitiu identificar configurações mais adequadas para o problema e obter uma distribuição mais representativa dos dados utilizados no treinamento.

Após a reformulação da abordagem e a reavaliação dos algoritmos, observou-se uma mudança no desempenho dos modelos. O LightGBM passou a apresentar o melhor equilíbrio entre acurácia, Macro F1-score e capacidade de classificação das diferentes classes, sendo selecionado como modelo final para as etapas posteriores do projeto.
